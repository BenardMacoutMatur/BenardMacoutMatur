```python
import streamlit as st
import sqlite3
from datetime import date

# ---- DATABASE SETUP ----
conn = sqlite3.connect('training.db')
cursor = conn.cursor()

cursor.execute('''
    CREATE TABLE IF NOT EXISTS training_log (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        date DATE NOT NULL,
        trained BOOLEAN NOT NULL,
        training_type TEXT,
        notes TEXT
    )
''')
conn.commit()

# ---- APP UI ----
st.title("🏋️‍♂️ Training Tracker App")

# Log training
st.subheader("Log Today's Training")

with st.form("log_form"):
    today = date.today()
    trained = st.radio("Did you train today?", ("Yes", "No"))
    training_type = st.text_input("Training type (e.g., Run, Gym, Yoga)", "")
    notes = st.text_area("Any notes?", "")
    submitted = st.form_submit_button("Submit")

    if submitted:
        cursor.execute('''
            INSERT INTO training_log (date, trained, training_type, notes)
            VALUES (?, ?, ?, ?)
        ''', (today, trained == "Yes", training_type, notes))
        conn.commit()
        st.success("✅ Training logged!")

# View history
st.subheader("📅 Training History")
rows = cursor.execute('SELECT * FROM training_log ORDER BY date DESC').fetchall()

for row in rows:
    st.write(f"**{row[1]}** - {'✅ Trained' if row[2] else '❌ No Training'} | {row[3]} | {row[4]}")

# Summary
st.subheader("📊 Summary Stats")
total_days = cursor.execute('SELECT COUNT(*) FROM training_log').fetchone()[0]
trained_days = cursor.execute('SELECT COUNT(*) FROM training_log WHERE trained = 1').fetchone()[0]

st.write(f"Total days logged: **{total_days}**")
st.write(f"Days you trained: **{trained_days}**")

conn.close()
```
